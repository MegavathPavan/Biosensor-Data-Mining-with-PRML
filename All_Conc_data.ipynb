```python
# prompt: train some unsupervised ml algorithms and plot the results for the data

from sklearn.cluster import KMeans, DBSCAN
from sklearn.decomposition import PCA
import numpy as np

# Assuming 'scaled_data' is your preprocessed data (scaled using StandardScaler)
# If not, uncomment the following lines and replace 'data' with your actual DataFrame
# scaler = StandardScaler()
# scaled_data = scaler.fit_transform(data)
# scaled_data = pd.DataFrame(scaled_data, columns=data.columns)


# KMeans Clustering
kmeans = KMeans(n_clusters=5, random_state=0) # You might need to tune the number of clusters
kmeans_labels = kmeans.fit_predict(scaled_data)
scaled_data['kmeans_labels'] = kmeans_labels

# DBSCAN Clustering
dbscan = DBSCAN(eps=0.5, min_samples=5)  # Adjust eps and min_samples as needed
dbscan_labels = dbscan.fit_predict(scaled_data)
scaled_data['dbscan_labels'] = dbscan_labels


# PCA for dimensionality reduction (optional, but helpful for visualization)
pca = PCA(n_components=2)
pca_result = pca.fit_transform(scaled_data.drop(['kmeans_labels', 'dbscan_labels', 'Conc'], axis=1, errors='ignore')) #Drop non-numeric columns

# Create a new DataFrame for PCA results
pca_df = pd.DataFrame(data=pca_result, columns=['PC1', 'PC2'])
pca_df['kmeans_labels'] = kmeans_labels
pca_df['dbscan_labels'] = dbscan_labels
pca_df['Conc'] = data['Conc']


# Plotting
plt.figure(figsize=(12, 6))

# KMeans plot
plt.subplot(1, 2, 1)
sns.scatterplot(x='PC1', y='PC2', hue='kmeans_labels', data=pca_df, palette='bright')
plt.title('KMeans Clustering (PCA)')

# DBSCAN plot
plt.subplot(1, 2, 2)
sns.scatterplot(x='PC1', y='PC2', hue='dbscan_labels', data=pca_df, palette='bright')
plt.title('DBSCAN Clustering (PCA)')

plt.tight_layout()
plt.show()


#Additional Plots (Optional) comparing clustering results to original data
plt.figure(figsize=(12, 6))
plt.subplot(1,2,1)
sns.scatterplot(x=' Vgs', y=' Ids', hue='kmeans_labels', data=scaled_data, palette='bright')
plt.title('KMeans Clustering (Vgs vs Ids)')
plt.subplot(1,2,2)
sns.scatterplot(x=' Vgs', y=' Ids', hue='dbscan_labels', data=scaled_data, palette='bright')
plt.title('DBSCAN Clustering (Vgs vs Ids)')
plt.tight_layout()
plt.show()
```

```python
inertia = []
k_range = range(1, 15)  # You can adjust the range based on your needs

for k in k_range:
    kmeans = KMeans(n_clusters=k, random_state=42)
    kmeans.fit(scaled_data)
    inertia.append(kmeans.inertia_)

# Plot the elbow curve
plt.figure(figsize=(8, 5))
plt.plot(k_range, inertia, marker='o')
plt.title('Elbow Method for Optimal k')
plt.xlabel('Number of Clusters (k)')
plt.ylabel('Inertia')
plt.show()
```

```python
from scipy.signal import savgol_filter
from numpy.polynomial.polynomial import Polynomial
from sklearn.linear_model import LinearRegression
from sklearn.preprocessing import PolynomialFeatures

features = df[[' Ids', ' Vgs', 'Conc']]
features_scaled = scaler.fit_transform(features)

features_scaled = pd.DataFrame(features_scaled, columns=[' Ids', ' Vgs', 'Conc'])

# Group data by 'conc' and apply smoothing
plt.figure(figsize=(8, 12))
for name, group in features_scaled.groupby('Conc'):
    vgs = group[' Vgs'].values
    ids = group[' Ids'].values

    # Sort data for consistent plotting
    sorted_indices = np.argsort(vgs)
    vgs = vgs[sorted_indices]
    ids = ids[sorted_indices]

    # Apply Savitzky-Golay filter to smooth data
    smoothed_ids = savgol_filter(ids, window_length=11, polyorder=2)

    # Plot original and smoothed
    plt.plot(vgs, ids, 'o', alpha=0.3, label=f'Original - {name}')
    plt.plot(vgs, smoothed_ids, '-', label=f'Smoothed - {name}')

plt.title("Smoothed IDS vs VGS")
plt.xlabel("VGS (Gate-Source Voltage)")
plt.ylabel("IDS (Current)")
plt.legend()
plt.show()
```

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from sklearn.cluster import KMeans
from numpy.polynomial.polynomial import Polynomial

# List of unique concentration labels (if available)
concentration_labels = df1['Conc'].unique()
print(concentration_labels)

# Feature extraction: fit a polynomial for each curve and extract coefficients
degree = 3  # Degree of polynomial (adjust based on curve complexity)
features = []

for label in concentration_labels:
    subset = df1[df1['Conc'] == label]
    # Fit a polynomial to Vgs vs Ids
    p = Polynomial.fit(subset[' Vgs'], subset[' Ids'], degree)
    # Store the polynomial coefficients as features for clustering
    features.append(p.convert().coef)

features = np.array(features)

# Apply K-Means on the polynomial coefficients
kmeans = KMeans(n_clusters=10, random_state=42)
clusters = kmeans.fit_predict(features)

# Assign cluster labels back to the concentrations
for i, label in enumerate(concentration_labels):
    df1.loc[df1['Conc'] == label, 'Cluster'] = clusters[i]

# Visualizing the clusters
plt.figure(figsize=(10, 6))
for cluster in range(10):
    cluster_data = df1[df1['Cluster'] == cluster]
    plt.plot(cluster_data[' Vgs'], cluster_data[' Ids'], label=f'Cluster {cluster}')

plt.title('K-Means Clustering of Vgs vs Ids Curves')
plt.xlabel('Vgs')
plt.ylabel('Ids')
plt.legend()
plt.show()
```
